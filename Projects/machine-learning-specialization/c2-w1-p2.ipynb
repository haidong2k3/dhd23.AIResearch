### What is Machine Learning?

* Định nghĩa chính thức đầu tiên của Machine Learning được đề xuất bởi Arthur Samuel năm 1595, khi nói về việc xây dựng một tiến trình máy tính có khả năng tự học hỏi để cải thiện khả năng chơi trò chơi Checker.  
    <q>Machine Learning is a field of study that gives computers the ability to learn without being explicitly programmed.</q>  
    \- Arthur Samuel (1959)-

* Phần lớn các thuật toán học máy (ML Algorithms) có thể được chia vào 3 nhóm chính Supervised Learning, Unsupervsided Learning và Reinforcement Learning.
* Trong đó, supervised learning là phổ biến nhất, đồng thời có tốc độ phát triển vô cùng mạnh mẽ. Supervised và Unsupervised đóng vai trò quan trọng trong hầu hết các bài toán ứng dụng hiện nay.
* Ngoài ra, cũng có nhiều thuật toán không thuộc nhóm trên - biến thể, phục vụ cho những bài toán cụ thể. (self-supervised, semi-supervised,...)


* Học tập, nghiên cứu về Machine Learning nói riêng và Aritificial Intelligent nói chung cần phải nắm vững 2 năng lực:
    - Learning: Toán học, logic phía sau các mô hình, thuật toán.
    - Applying: khả năng ứng dụng và điều chỉnh trong các bài toán thực tế.

### Supervised Learning

* Supervised Learning là thuật toán học dựa trên dữ liệu đã được gán nhãn, hướng đến việc tìm "câu trả lời chính xác" đối với mỗi đầu vào.
    ```
    X (input) -> Y (output label)
    ```

* Với mỗi điểm dữ liệu x, sẽ có label y là "correct answer" tương ứng. Sau khi học từ bộ dữ liệu này, mô hình sẽ có khả năng dự đoán label của các điểm dữ liệu khác dù chưa từng được học với độ chính xác nhất định.
    ```
    (data point, correct answer) -> model learning -> prediction
    ```
    
* Một số bài toán ứng dụng phổ biến như:
    - spam filtering: email -> spam/not spam
    - speech recognition: sound/audio -> text/script
    - language translation: original language -> target language
    - online advertising: user/products -> suitable
    - self-driving car: images/radar -> obstacles position
    - ...

* Phần lớn các bài toán thuộc Supervised Learning đều có thể phân vào 2 nhóm chính là regression và classification.
* Regression hay bài toán hồi quy, thường là dự đoán giá trị trong 1 trường liên tục: 
    - predict a number
    - infinitely many possible outputs
    - ex: House price prediction
* Classification hay bài toán phân loại, phân loại dữ liệu vào các nhóm đã có sẵn từ trước: 
    - predict categories
    - small number of possible outputs
    - ex: breast cancer detection

=> Tuỳ vào trường hợp cụ thể, các thuật toán regresion cũng có thể giải quyết bài toán classification và ngược lại.

### Unsupervised Learning

* Unsupervised Learning có thể xem như là một thuật toán "khai thác" dữ liệu, dữ liệu đầu vào không được gán nhãn hay một "kết quả chính xác". Hướng đến việc tìm ra những cấu trúc, kiến thức ẩn bên trong dữ liệu.
    ```
    X (input) -> structure, information of X
    ```

* Bài toán trọng tâm thuộc nhóm này là bài toán Clustering
    - Group similar data points together
    - ex: cluster related news; DNA microarray; group customers

* Một số bài toán khác cũng được xem là unsupervised learning như:
    - anomaly detection: find unusual data points
    - dimentionality reduction: compress data using fewer numbers
